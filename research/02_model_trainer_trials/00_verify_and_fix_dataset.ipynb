```python
#import libraries
import os
import random
import matplotlib.pyplot as plt
from PIL import Image
```

```python
#set path for dataset

BASE_DIR = r"C:\Users\sagal\Desktop\Let us build\emotion_detection\dataset\Organized"
TRAIN_DIR = os.path.join(BASE_DIR, "train")

print(f"Target path: {TRAIN_DIR}")
print(f"Directory exists? {os.path.exists(TRAIN_DIR)}")
```

```python
classes = [c for c in os.listdir(TRAIN_DIR) if os.path.isdir(os.path.join(TRAIN_DIR, c))]
classes.sort()

#retrieve names
```

```python
num_classes = len(classes)
print(f"Found {num_classes} subfolders: {classes}")
```

```python
# Plot 5 random samples per class
fig, axes = plt.subplots(num_classes, 5, figsize=(15, 3 * num_classes))

for row_idx, cls in enumerate(classes):
    cls_folder = os.path.join(TRAIN_DIR, cls)
    images = [f for f in os.listdir(cls_folder) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
    
    # Pick up to 5 random images
    sampled_imgs = random.sample(images, min(5, len(images)))
    
    for col_idx, img_name in enumerate(sampled_imgs):
        img_path = os.path.join(cls_folder, img_name)
        img = Image.open(img_path)
        
        ax = axes[row_idx, col_idx] if num_classes > 1 else axes[col_idx]
        ax.imshow(img)
        ax.set_title(f"Folder: '{cls}'", fontsize=11, fontweight='bold')
        ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
import os

BASE_DIR = r"C:\Users\Desktop\Let us build\emotion_detection\dataset\Organized"

# Map your folder names on the left to standardized numbered prefixes on the right
# (Adjust the keys on the left if your folders are currently numbers like '1', '2' or strings)
RENAME_MAP = {
    "surprise": "0_surprise",  # or "1": "0_surprise"
    "fear":     "1_fear",      # or "2": "1_fear"
    "disgust":  "2_disgust",   # or "3": "2_disgust"
    "happy":    "3_happiness", # or "4": "3_happiness"
    "sad":      "4_sadness",   # or "5": "4_sadness"
    "angry":    "5_anger",     # or "6": "5_anger"
    "neutral":  "6_neutral"    # or "7": "6_neutral"
}

for split in ["train", "test"]:
    split_path = os.path.join(BASE_DIR, split)
    if not os.path.exists(split_path):
        continue
        
    print(f"\n--- Processing split: {split} ---")
    for old_name, new_name in RENAME_MAP.items():
        old_dir = os.path.join(split_path, old_name)
        new_dir = os.path.join(split_path, new_name)
        
        if os.path.exists(old_dir):
            os.rename(old_dir, new_dir)
            print(f"Renamed: {old_name} -> {new_name}")
        else:
            print(f"Skipped (Not Found): {old_name}")

print("\nDataset directories successfully renamed!")
```
